# SBI2 hackathon: download the participant datasets

Use this notebook to choose, preview, and download the processed datasets for the hackathon. It follows the Altair starter notebook workflow: first inspect what will be downloaded, then run the download, then inspect the resulting profiles.

Before opening the notebook, install the repository environment with `python -m pip install -r requirements.txt` from the repository root.

## Available datasets

| Selection | Contents | Approximate size |
|---|---|---:|
| `altair-phase` | A549 phase profiles per guide and per guide-by-screen, plus small metadata | 45.8 MiB + 2.92 GiB + metadata |
| `altair-4i` | Altair 4i marker archive, with 10 Biohub_OPS0077 guide-by-screen profiles extracted | 91.9 MiB extracted; requires downloading the 1.90 GiB marker archive |
| `posh` | POSH published 124-gene morphology profile table | 4.47 MiB |
| `aconcagua` | HeLa interphase and mitotic aggregate profiles plus metadata | 12.1 MiB |
| Supplied Aconcagua guide/gene profiles | Six guide- and gene-level Parquet tables across all-cell, interphase, and mitotic populations | 624.1 MiB, staged from a local folder |

The four-group core download is approximately 5.24 GB (4.88 GiB) of network transfer, excluding the separately supplied Aconcagua guide/gene tables (about 624 MiB local copy), mainly `phase_guide_x_screen.h5ad` (3.13 GB) and `markers_guide.zip` (2.04 GB), plus small metadata files. The 4i guide profiles are extracted from `markers_guide.zip`; the ZIP has no standalone member URLs. The optional sampled cell-level 4i files are much larger and are not included. Aconcagua `all` is not published as a processed profile, so both available populations are downloaded. No raw images or Aconcagua cell-level Parquet are included.

POSH data are published under CC BY-NC-SA 4.0. Confirm the source terms for the event and any redistribution. Keep downloaded data out of GitHub.

````python
from pathlib import Path
import os
import subprocess
import sys

HERE = Path.cwd().resolve()
ROOT = next((p for p in [HERE, *HERE.parents] if (p / 'scripts' / 'download_data.py').is_file()), None)
if ROOT is None:
    raise FileNotFoundError('Open this notebook from the sbi2-hackathon-prep-data-ops repository.')

# On the VM, set SBI2_DATA_DIR=/mnt/hackathon-data before starting Jupyter.
DATA_DIR = Path(os.environ.get('SBI2_DATA_DIR', ROOT / 'data')).expanduser().resolve()
ACON_GUIDE_SOURCE = Path(os.environ.get('ACONCAGUA_GUIDE_SOURCE_DIR', ROOT.parent / 'aconcagua_guide_gene_features')).expanduser().resolve()
STAGE_ACONCAGUA_GUIDE_DATA = False
DATASET_CHOICES = ('altair-phase', 'altair-4i', 'posh', 'aconcagua')
DATASETS_TO_DOWNLOAD = ['altair-phase', 'altair-4i', 'posh', 'aconcagua']

# Set to True in the download cell below only after reviewing the size preview.
DOWNLOAD_NOW = False
print('Repository:', ROOT)
print('Data directory:', DATA_DIR)
print('Selected:', DATASETS_TO_DOWNLOAD)
print('Aconcagua guide/gene source:', ACON_GUIDE_SOURCE)
print('Source folder found:', ACON_GUIDE_SOURCE.is_dir())
````

## 1. Preview the selected download

Edit `DATASETS_TO_DOWNLOAD` above to select only what you need. This cell queries current source sizes and writes no data files. The Altair phase group includes both guide-level summaries. Keep `altair-4i` selected for Biohub_OPS0077 4i profiles; preview the full marker-archive transfer size before downloading.

````python
unknown = set(DATASETS_TO_DOWNLOAD) - set(DATASET_CHOICES)
if unknown:
    raise ValueError(f'Unknown dataset selections: {sorted(unknown)}')
if not DATASETS_TO_DOWNLOAD:
    raise ValueError('Select at least one dataset group.')

download_script = ROOT / 'scripts' / 'download_data.py'
preview_cmd = [sys.executable, str(download_script), '--datasets', *DATASETS_TO_DOWNLOAD, '--data-dir', str(DATA_DIR), '--dry-run']
preview = subprocess.run(preview_cmd, cwd=ROOT, text=True, capture_output=True)
print(preview.stdout)
if preview.stderr:
    print(preview.stderr)
if preview.returncode:
    raise RuntimeError(f'Download preview failed with exit code {preview.returncode}')
````

## 2. Download the selected data

After reviewing the preview, set `DOWNLOAD_NOW = True` in the setup cell and run this cell. The downloader verifies known SHA-256 hashes, checks object sizes, writes files atomically, extracts the 10 OPS0077 profiles, and saves `download_manifest.json`. Altair phase downloads include both `phase_guide.h5ad` and `phase_guide_x_screen.h5ad`. Re-running it reuses verified existing files.

````python
if DOWNLOAD_NOW:
    download_cmd = [sys.executable, str(download_script), '--datasets', *DATASETS_TO_DOWNLOAD, '--data-dir', str(DATA_DIR), '--yes']
    subprocess.run(download_cmd, cwd=ROOT, check=True)
else:
    print('Download skipped. Review the preview, set DOWNLOAD_NOW = True, then rerun this cell.')
````

## 3. Stage the supplied Aconcagua guide/gene profiles

The six Parquet files are provided locally in `aconcagua_guide_gene_features/` and are not hosted at a URL configured for this downloader. Transfer that folder to the VM, set `ACONCAGUA_GUIDE_SOURCE_DIR` if needed, review the checksum-verified dry run below, then set `STAGE_ACONCAGUA_GUIDE_DATA = True` to copy them into the selected data directory.

````python
stage_script = ROOT / 'scripts' / 'stage_aconcagua_guide_data.py'
if ACON_GUIDE_SOURCE.is_dir():
    stage_preview = subprocess.run([sys.executable, str(stage_script), '--source-dir', str(ACON_GUIDE_SOURCE), '--data-dir', str(DATA_DIR), '--dry-run'], cwd=ROOT, text=True, capture_output=True)
    print(stage_preview.stdout)
    if stage_preview.stderr:
        print(stage_preview.stderr)
    if stage_preview.returncode:
        raise RuntimeError(f'Aconcagua guide-data preflight failed with exit code {stage_preview.returncode}')
else:
    print('Bundle not found. Transfer it to the VM and set ACONCAGUA_GUIDE_SOURCE_DIR.')
````

````python
if STAGE_ACONCAGUA_GUIDE_DATA:
    if not ACON_GUIDE_SOURCE.is_dir():
        raise FileNotFoundError(f'Aconcagua guide-data source folder not found: {ACON_GUIDE_SOURCE}')
    subprocess.run([sys.executable, str(stage_script), '--source-dir', str(ACON_GUIDE_SOURCE), '--data-dir', str(DATA_DIR), '--yes'], cwd=ROOT, check=True)
else:
    print('Staging skipped. Set STAGE_ACONCAGUA_GUIDE_DATA = True to copy the verified files.')
````

## 4. Review files and run first-pass quality checks

The quality report checks file hashes, H5AD and Parquet readability, profile dimensions, duplicate identifiers, metadata columns, and guide/target overlap between each OPS0077 profile and both Altair phase summaries when present. Missing groups are listed when you intentionally download only a subset. Aconcagua guide IDs are checked within that dataset; cross-atlas overlap is calculated by gene symbol, not by assuming guide barcode and guide sequence IDs are interchangeable. These checks do not establish biological validity or independently verify the originating guide library.

````python
manifest_path = DATA_DIR / 'download_manifest.json'
if manifest_path.is_file():
    import json
    manifest = json.loads(manifest_path.read_text())
    for item in manifest['files']:
        print(f"{item['size_bytes'] / 1024**2:8.2f} MiB  {item['path']}  sha256={item['sha256'][:12]}…")
else:
    print('No remote-download manifest yet; QC will still inspect any locally staged files.')

qc_cmd = [sys.executable, str(ROOT / 'scripts' / 'run_quality_checks.py'), '--data-dir', str(DATA_DIR), '--outdir', str(DATA_DIR / 'qc'), '--allow-incomplete']
qc = subprocess.run(qc_cmd, cwd=ROOT, text=True, capture_output=True)
print(qc.stdout)
if qc.stderr:
    print(qc.stderr)
if qc.returncode:
    raise RuntimeError(f'Quality checks reported an error; inspect {DATA_DIR / "qc" / "quality_report.json"}')
````
